## Configuration and Imports

````python
from datasets import load_dataset
from tqdm import tqdm

# --- CONFIGURATION ---
DATASET_PATH = "squad_v2"
DATASET_SPLIT = "train"
SAMPLE_SIZE = 130000

CODE_KEYWORDS = {
    "def ",
    "class ",
    "import ",
    "return ",
    "print(",
    "function",
    "python",
    "javascript",
    "html",
    "css",
    "c++",
    "java",
    "sql",
    "compile",
    "debug",
    "variable",
    "array",
    "syntax",
    "algorithm",
    "```",
}
MATH_KEYWORDS = {
    "calculate",
    "equation",
    "theorem",
    "integral",
    "derivative",
    "algebra",
    "geometry",
    "math",
    "\\frac",
    "\\sqrt",
    "sum of",
}
MED_KEYWORDS = {
    "symptom",
    "diagnosis",
    "treatment",
    "patient",
    "syndrome",
    "mg ",
    "dosage",
    "clinical",
    "doctor",
    "disease",
    "infection",
}
````

## EDA

````python
def analyze_dataset_contamination():
    print(f"\nStreaming up to {SAMPLE_SIZE} prompts from {DATASET_PATH}...")
    ds = load_dataset(
        DATASET_PATH, split=DATASET_SPLIT, streaming=True, trust_remote_code=True
    )

    stats = {
        "total_analyzed": 0,
        "contains_code": 0,
        "contains_math": 0,
        "contains_med": 0,
        "clean_general": 0,
    }

    for row in tqdm(ds, total=SAMPLE_SIZE, desc=f"Analyzing {DATASET_PATH}"):
        if stats["total_analyzed"] >= SAMPLE_SIZE:
            break

        # Handle different dataset structures based on config
        prompt = ""
        if DATASET_PATH == "squad_v2":
            prompt = row.get("question", "")
        elif DATASET_PATH == "daily_dialog":
            dialog = row.get("dialog", [])
            if dialog:
                prompt = dialog[0]  # Grab the first utterance

        if not prompt:
            continue

        prompt = prompt.lower()
        stats["total_analyzed"] += 1

        # Check contamination
        has_code = any(kw in prompt for kw in CODE_KEYWORDS)
        has_math = any(kw in prompt for kw in MATH_KEYWORDS)
        has_med = any(kw in prompt for kw in MED_KEYWORDS)

        if has_code:
            stats["contains_code"] += 1
        if has_math:
            stats["contains_math"] += 1
        if has_med:
            stats["contains_med"] += 1

        if not (has_code or has_math or has_med):
            stats["clean_general"] += 1

    # Print Report
    print("\n" + "=" * 50)
    print(f"{DATASET_PATH.upper()} CONTAMINATION REPORT")
    print("=" * 50)
    total = stats["total_analyzed"]

    if total == 0:
        print("No prompts analyzed.")
        return

    print(f"Total Prompts Analyzed: {total:,}")
    print("-" * 50)

    code_pct = (stats["contains_code"] / total) * 100
    math_pct = (stats["contains_math"] / total) * 100
    med_pct = (stats["contains_med"] / total) * 100
    clean_pct = (stats["clean_general"] / total) * 100

    print(
        f"Contains CODE keywords:    {stats['contains_code']:>5,} ({code_pct:>5.1f}%)"
    )
    print(
        f"Contains MATH keywords:    {stats['contains_math']:>5,} ({math_pct:>5.1f}%)"
    )
    print(f"Contains MEDICAL keywords: {stats['contains_med']:>5,} ({med_pct:>5.1f}%)")
    print("-" * 50)
    print(
        f"CLEAN (No target keywords):{stats['clean_general']:>5,} ({clean_pct:>5.1f}%)"
    )
    print("=" * 50)
````

## Execution

````python
if __name__ == "__main__":
    analyze_dataset_contamination()
````
